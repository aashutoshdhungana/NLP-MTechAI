```python
cat = [0.4, 0.2, 0.8]
dog = [0.5, 0.1, 0.7]
```

```python
dot_product = 0
for item in zip(cat, dog):
    dot_product += item[0] * item[1]
print(dot_product)
```

```python
mag_a = sum(x**2 for x in cat) ** 0.5
mag_b = sum(x**2 for x in dog) ** 0.5
print(mag_a, mag_b)
```

```python
cosine_similarity = dot_product / (mag_a * mag_b)
print(cosine_similarity)
```

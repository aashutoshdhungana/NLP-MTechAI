```python
word_counts = {
    "cat": 4,
    "dog": 3,
    "bird": 1
}
```

```python
total_count = sum(word_counts.values())
total_count
```

```python
vocab_size = len(word_counts.keys())
vocab_size
```

```python
probabilities = {key: (word_counts[key] + 1) / (total_count + vocab_size) for key in word_counts.keys()}
```

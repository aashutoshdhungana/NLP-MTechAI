```python
sentence = "the cat sat on the mat"
```

```python
window_size = 1
```

```python
all_words = sentence.split()

unique_words = list(dict.fromkeys(all_words))

word_to_idx = {word:i for i, word in enumerate(unique_words)}
idx_to_word = {i:word for i, word in enumerate(unique_words)}

word_counts = len(unique_words)
```

```python
co_occurance_mat = [[0 for _ in  range(word_counts)] for _ in range(word_counts) ]
```

```python
for i, target_word in enumerate(all_words):
    target_idx = word_to_idx[target_word]

    for j in range(1, window_size + 1):
        if i + j < len(all_words):
            context_word = all_words[i + j]
            context_idx = word_to_idx[context_word]

            # Increment symmetric co-occurrence count
            co_occurance_mat[target_idx][context_idx] += 1
            co_occurance_mat[context_idx][target_idx] += 1

print(co_occurance_mat)
```

```python
embeddings = {
"cat":[0.4,0.2,0.8],
"dog":[0.5,0.1,0.7],
"car":[-0.7,0.8,-0.2],
"lion":[0.45,0.25,0.82]
}
```

```python
def dot_product(a, b):
    return sum(x*y for x, y in zip(a, b))
```

```python
def magnitude(v):
    return sum(x**2 for x in v) ** 0.5
```

```python
def cosine_similarity(a, b):
    dot_product_value = dot_product(a, b)
    mag_a = magnitude(a)
    mag_b = magnitude(b)
    return dot_product_value / (mag_a * mag_b)
```

```python
similarities = {}
for word1 in embeddings:
    for word2 in embeddings:
        if word1 != word2:
            sim = cosine_similarity(embeddings[word1], embeddings[word2])
            similarities[(word1, word2)] = sim
```

```python
max_similarity = max(similarities, key=similarities.get)
```

```python
print(f"The most similar words are '{max_similarity[0]}' and '{max_similarity[1]}' with a cosine similarity of {similarities[max_similarity]:.4f}.")
```

```python
courpus = [
    "I like NLP",
    "I like AI",
    "AI is amazing"
]
```

```python
all_words = []
```

```python
for text in courpus:
    words = text.split()
    all_words.extend(words)
```

```python
vocabulary = set(all_words)
print("The vocabulary of the courpus is:")
print(vocabulary)
print(f"The vocabulary size is {len(vocabulary)}")
```

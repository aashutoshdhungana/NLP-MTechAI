```python
pair_counts = {}
single_counts = {}
for text in training_text:
    words = text.split()
    for i in range(len(words) - 1):
        w1, w2 = words[i], words[i+1]
        pair = (w1,w2) 
        if pair in pair_counts.keys():
            pair_counts[pair] += 1
        else:
            pair_counts[pair] = 1

        if w1 in single_counts.keys():
            single_counts[w1] += 1
        else:
            single_counts[w1] = 1
```

```python
prediction_text = "mero"
```

```python
prediction_words = prediction_text.split()
base_word = prediction_words[-1]
```

```python
for i in range(5):
    matching_pairs = {pair: count for pair, count in pair_counts.items() if pair[0] == base_word}
    
    if not matching_pairs:
        break

    top_3_pairs = sorted(matching_pairs.items(), key=lambda item: item[1], reverse=True)[:3]
    top_3_predictions = [(pair[1], count) for pair, count in top_3_pairs]
    print(f"Top predictions for '{base_word}':", top_3_predictions)
    predicted_word = top_3_predictions[0][0]
    prediction_text += " " + predicted_word
    base_word = predicted_word
```

```python
documents = [
    'i like nlp',
    'i like ai',
    'ai is amazing'
]
```

```python
terms = ' '.join(documents).split()
```

```python
from collections import defaultdict
document_frequency = defaultdict(int)
tf = []
idf = defaultdict(float)
tf_idf = []
```

```python
for document in documents:
    doc_terms = document.split()
    for term in terms:
        if term in doc_terms:
            document_frequency[term] += 1
```

```python
import math
for term, doc_count in document_frequency.items():
    idf[term] = math.log10(len(documents) / doc_count)
```

```python
print("DF:",document_frequency)
```

```python
print("IDF", idf)
```

```python
for document in documents:
    document_terms = document.split()
    total_terms = len(document_terms)
    term_frequency = defaultdict(float)
    tf_idf_doc = defaultdict(float)
    for term in document_terms:
        term_frequency[term] += 1

    for item, count in term_frequency.items():
        term_frequency[item] =  count / total_terms

    for term in document_terms:
        tf_idf_doc[term] = term_frequency[term]  * idf[term]
    tf.append(term_frequency)
    tf_idf.append(tf_idf_doc)
```

```python
for i, item in enumerate(tf):
    print()
    print(f"The term frequency for document {i+1} is:")
    print()
    print(f"{"Term":15}|{"Frequency":15}")
    for item, count in tf[i].items():
        print(f"{item:15}|{count:15.2f}")
```

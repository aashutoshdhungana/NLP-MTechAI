# Reading the article

```python
article = ""
with open('article.txt', 'r') as file:
    article = file.read()
article
```

# Importing required packages

```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(article)
doc
```

# Printing the named entities

```python
for entity in doc.ents:
    print(f"Entity: {entity.text}, Label: {entity.label_}")
```

# Saving the named entities in csv file

```python
import csv

with open("AashutoshDhungana_NER_01.csv", "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Entity", "Entity_Type"])
    for entity in doc.ents:
        writer.writerow([entity.text, entity.label_])
```

```python
labels = [entity.label_ for entity in doc.ents]
set(labels)
```

# Different types of entity types in the article

- Cardinal: Cardinals are the numerals that do not fall under any other types e.g. 3, hunderd etc
- Date: Absolute and Relative dates and periods e.g. 2026, yesterday
- GPE: Geopolitical entities like countries, cities, states, provinces e.g. Nepal, Kathmandu etc
- LAW: Named legal documents. E.g. Tourism Bill
- LOC: Non-GPE locations like mountain ranges, oceans etc
- NORP: Nationalities, Religious groups, Political group etc.
- ORDINAL: Words indicating an order or position e.g first second, 9th
- ORG: Organizations, companies etc.
- PERSON: Fictional or real people including full names, first or lastnames
- Quantity: Measurements like, 5km 1kg etc
- WORK_OF_ART: Titles of books, songs, paintings etc.

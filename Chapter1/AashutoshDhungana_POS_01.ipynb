# Reading text from the file

```python
article = ""
with open('article.txt', 'r') as file:
    article = file.read()

print(article)
```

# Importing NLTK and downloading essential data packs

```python
import nltk

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt_tab')
```

# Tokenizing the article

```python
tokens = nltk.word_tokenize(article)
tokens
```

# Performing POS on all the tokens

```python
pos_tags = nltk.pos_tag(tokens)
pos_tags
```

# Extacting Nouns

```python
nount_tags = { 'NN', 'NNS', 'NNP', 'NNPS' }
nouns = [word for word, tag in pos_tags if tag in nount_tags]
nouns
```

# Extracting verbs

```python
verbs = [word for word, tag in pos_tags if tag.startswith('VB')]
verbs
```

# Saving the words along with the POS tags into a CSV file

```python
import csv
```

```python
no_of_rows = max(len(nouns), len(verbs))
```

```python
with open("AashutoshDhungana_POS_Identifier.csv", "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Word", "Tags"])
    for word, tag in pos_tags:
        writer.writerow([word, tag])
```
